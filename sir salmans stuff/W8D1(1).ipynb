```python
import tensorflow as tf
```

```python
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.imdb.load_data(num_words=1000)
```

```python
x_seq = tf.keras.preprocessing.sequence.pad_sequences(x_train, maxlen=1000)
x_seq_test = tf.keras.preprocessing.sequence.pad_sequences(x_test, maxlen=1000)
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Embedding(1000, 16),
    tf.keras.layers.SimpleRNN(32, activation='relu', name='first_rnn', return_sequences=True),
    tf.keras.layers.SimpleRNN(32, activation='relu', name='second_rnn', return_sequences=True),
     tf.keras.layers.SimpleRNN(32, activation='relu', name='third_rnn'),
    tf.keras.layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(x_seq, y_train, epochs=10, validation_split=.2, batch_size=32)
```

```python
model.evaluate(x_seq_test, y_test)
```
